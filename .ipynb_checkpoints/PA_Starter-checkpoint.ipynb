```python
import warnings; warnings.filterwarnings("ignore")
import mlflow, numpy as np, pandas as pd

NAME = "CHUA HANG JIE"      # e.g. "AISYAH BINTI AHMAD"
L4   = "0511"                # last four digits of your IC, keep the leading zero

# ---- your personal values (rules in the paper) ----
SEED      = int(L4)
digits = [int(d) for d in L4]
TEST_SIZE = 0.20 if digits[3] % 2 == 0 else 0.25
DEPTH = 3 + (digits[1] % 4)
C1        = 0.08
C2        = C1 * 10
RANK      = f"test_recall"
PORT      = 5111
EXP       = f"LangganKu_Churn_0511"
REG       = f"langganku_churn_0511"
```

## Task 1 — Set-up, split and pipeline

```python
# your code
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

mlflow.set_tracking_uri(f'sqlite:///langganku_0511.db')
mlflow.set_experiment(EXP)
print('Fracking URI:', mlflow.get_tracking_uri())

df = pd.read_csv('subscriber_churn.csv')
print(df.shape)
display(df.head())
```

```python
X= df.drop(columns=['subscriber_id', 'churned'])
y = df['churned']
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=TEST_SIZE,
    random_state=SEED
)

cat_cols = X.select_dtypes(exclude='number').columns.tolist()
num_cols = X.select_dtypes(include='number').columns.tolist()
print("categorical:", cat_cols); print("numeric:", num_cols)
print("train / test rows:", len(X_train), len(X_test))

def make_pipeline(model):
    pre = ColumnTransformer([('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)],
                           remainder='passthrough')
    return Pipeline([('prep', pre), ("model", model)])
```

## Task 2 — Track the candidates in ONE experiment

```python
# your code
import mlflow.sklearn
from mlflow.models import infer_signature
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score,recall_score, f1_score

def get_metrics(y_true, y_pred):
    return {
        'accuracy': float(accuracy_score(y_true, y_pred)),
        'precision': float(precision_score(y_true, y_pred, zero_division=0)),
        "recall": float(recall_score(y_true, y_pred)),
        'f1': float(f1_score(y_true, y_pred))
    }

candidates = [
    ('logreg-base', LogisticRegression(max_iter=1000),{'max_iter': 1000}, 'linear'),
    (f"logreg-c{C1:g}", LogisticRegression(C=C1, max_iter=1000),{"C1":C1, "max_iter":1000}, 'linear'),
    (f"logreg-c{C2:g}", LogisticRegression(C=C2, max_iter=1000),{"C2":C2, "max_iter":1000}, 'linear'),
    (f"rf-d{DEPTH}", RandomForestClassifier(n_estimators=100, max_depth=DEPTH, random_state=SEED),
    {'n_estimators':100, "max_depth": DEPTH}, 'tree_ensemble')
]

run_ids = {}
for prefix, model, hparams, mtype in candidates:
    run_name = f"{prefix}-{L4}"
    with mlflow.start_run(run_name=run_name) as run:
        pipe = make_pipeline(model).fit(X_train, y_train)

        train_m = get_metrics(y_train, pipe.predict(X_train))
        test_m = get_metrics(y_test, pipe.predict(X_test))

        mlflow.log_params({
            'test_size': TEST_SIZE,
            'split_seed': SEED,
            'model': type(model).__name__, **hparams
        })
        mlflow.set_tags({
            'model_type': mtype,
            'candidate': NAME
        })
        mlflow.log_metrics({f"train_{k}": v for k, v in train_m.items()})
        mlflow.log_metrics({f'test_{k}': v for k, v in test_m.items()})

        sig = infer_signature(X_train, pipe.predict(X_train))
        mlflow.sklearn.log_model(
            pipe,
            name='model',
            signature=sig,
            input_example=X_train.head(3),
            skops_trusted_types=['sklearn.tree._tree.Tree']
        )

        run_ids[run_name] = run.info.run_id
        print(f"{run_name:<22} train={train_m}\n{'':<22} test={test_m}")
```

## Task 3 — Choose the best run from code and justify it

```python
#### your code
runs = mlflow.search_runs(
    experiment_names=[EXP],
    filter_string='attributes.status = "FINISHED"',
    order_by=["metrics.test_recall DESC"]
)
runs['gap_recall'] = (runs['metrics.train_recall'] - runs['metrics.test_recall']).abs()
display(runs[[
    'tags.mlflow.runName',
    'tags.model_type',
    'metrics.test_f1',
    'metrics.test_recall',
    'metrics.test_precision',
    'metrics.test_accuracy',
    'metrics.train_f1',
    'gap_recall'
]])

best = runs.iloc[0]
BEST_RUN_ID = best['run_id']
BEST_NAME = best['tags.mlflow.runName']
BASE_NAME = candidates[0][0] + f"-{L4}"
BASE_RUN_ID = run_ids[BASE_NAME]
print(f"Best run: {BEST_NAME}  (test_recall = {best['metrics.test_recall']:.4f})")
```

**Why I chose this run (2–3 sentences):**
- I ranked the runs from code by **test_recall** (Higher is better).
- Model run name ``logreg-c0.8-0511`` came first when I arrange them in decending order (Higher to lowest), with the **test_recall** score around **0.4930** (4 decimal place). Based on the data, it's value is identical to run name ``logreg-base-0511``. But the reason I choose the first one is because the ``gap_recall`` is smaller. 

## Task 4 — Register, alias and roll back

```python
# your code
from mlflow import MlflowClient
client = MlflowClient()

v1 = mlflow.register_model(f"runs:/{BASE_RUN_ID}/model", REG)
client.set_registered_model_alias(REG, 'champion', v1.version)
print('v1 =', BASE_NAME, '-> champion')


v2 = mlflow.register_model(f"runs:/{BEST_RUN_ID}/model",REG)
client.set_registered_model_alias(REG, 'champion', v2.version)
client.set_model_version_tag(REG, v2.version, 'approved_by', NAME)
client.update_model_version(
    REG, v2.version,
    description=f"{BEST_NAME}: best test_recall = {best['metrics.test_recall']:.4f} on the held-out test set "
                f"(split seed {SEED}, test_size {TEST_SIZE}). Promoted over baseline v{v1.version}"
)
print('v2 = ', BEST_NAME, '-> champion')


client.set_registered_model_alias(REG, 'champion', v1.version)
print('ROLLACK champion ->', client.get_model_version_by_alias(REG, 'champion').version)


client.set_registered_model_alias(REG, 'champion', v2.version)
mv = client.get_model_version_by_alias(REG, 'champion')
print(f'FINAL    champion -> v{mv.version} | run {mv.run_id} | tags {mv.tags}')
```

## Task 5 — Serve the champion and call it

```python
# your code
import os, sys, time, subprocess, requests

MODEL_URI = f"models:/{REG}@champion"
URL = f"http://127.0.0.1:{PORT}"

def server_up():
    try:    return requests.get(URL + "/ping", timeout=2).status_code == 200
    except requests.exceptions.RequestException: return False

server = None
if not server_up():
    env = {**os.environ, "MLFLOW_TRACKING_URI": f"sqlite:///langganku_{L4}.db"}
    server = subprocess.Popen([sys.executable, "-m", "mlflow", "models", "serve", "-m", MODEL_URI,
                               "-p", str(PORT), "--env-manager", "local"], env=env)
    for _ in range(90):                     # wait up to ~90 s for the server to start
        if server_up(): break
        time.sleep(1)
print("server up:", server_up(), "on", URL)
```

```python
# "Your row": row number L4 mod n of the feature table (the same row the checker uses)
row = X.iloc[[int(L4) % len(X)]]
display(row)

payload = {"dataframe_split": row.to_dict(orient="split", index=False)}
resp = requests.post(URL + "/invocations", json=payload, timeout=30)
print("HTTP", resp.status_code, resp.json())

pred = resp.json()["predictions"][0]
local = mlflow.pyfunc.load_model(MODEL_URI).predict(row)[0]
print(f"server predicted class (1 = will default): {pred}   | same model loaded locally: {local}")
```
